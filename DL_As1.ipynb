```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings
import math
warnings.filterwarnings('ignore')

# Deep Learning
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.applications import ResNet50, EfficientNetB0
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras import mixed_precision

# Metrics
from sklearn.metrics import (accuracy_score, f1_score, cohen_kappa_score,
                            roc_auc_score, precision_recall_curve, auc,
                            confusion_matrix, classification_report)
from scipy.stats import pearsonr

# ============================================================================
# GPU CONFIGURATION
# ============================================================================
print("="*80)
print("GPU CONFIGURATION")
print("="*80)

# Enable memory growth for GPU
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
        print(f"✓ GPU Found: {gpus}")
        print(f"✓ Memory growth enabled")

        # Enable mixed precision for better performance on RTX 3050
        mixed_precision.set_global_policy('mixed_float16')
        print(f"✓ Mixed precision enabled (float16)")
    except RuntimeError as e:
        print(f"GPU setup error: {e}")
else:
    print("⚠️ No GPU found, using CPU")

# Set random seeds
np.random.seed(42)
tf.random.set_seed(42)

print(f"\nTensorFlow version: {tf.__version__}")


# ============================================================================
# DATA LOADING MODULE
# ============================================================================

class DataLoader:
    """Handles loading and preprocessing of facial expression dataset"""

    def __init__(self, data_path):
        self.data_path = Path(data_path)
        self.img_size = (224, 224)
        self.emotion_labels = {
            0: 'Neutral', 1: 'Happy', 2: 'Sad', 3: 'Surprise',
            4: 'Fear', 5: 'Disgust', 6: 'Anger', 7: 'Contempt'
        }

    def load_annotations(self, split='train'):
        """Load annotations from individual files for each image"""
        ann_folder = self.data_path / 'annotations'

        # Get all annotation files
        exp_files = sorted([f for f in os.listdir(ann_folder) if f.endswith('_exp.npy')])

        expressions = []
        valence = []
        arousal = []

        print(f"  Loading annotations from {len(exp_files)} files...")

        for exp_file in exp_files:
            # Get the image number from filename (e.g., "0_exp.npy" -> "0")
            img_num = exp_file.split('_')[0]

            # Load corresponding annotation files
            try:
                exp_val = np.load(ann_folder / f'{img_num}_exp.npy', allow_pickle=True).item()
                val_val = np.load(ann_folder / f'{img_num}_val.npy', allow_pickle=True).item()
                ar_val = np.load(ann_folder / f'{img_num}_aro.npy', allow_pickle=True).item()

                expressions.append(exp_val)
                valence.append(val_val)
                arousal.append(ar_val)
            except:
                continue

        annotations = np.column_stack([
            np.array(expressions),
            np.array(valence),
            np.array(arousal)
        ])

        print(f"  Loaded {len(annotations)} annotations")
        return annotations

    def load_images(self, split='train'):
        """Load all images (split will be done later)"""
        img_folder = self.data_path / 'images'

        # Get all image files sorted by number
        img_files = []
        for f in os.listdir(img_folder):
            if not f.startswith('.'):
                try:
                    num = int(f.split('.')[0]) if '.' in f else int(f)
                    img_files.append((num, f))
                except:
                    continue

        img_files.sort()
        print(f"  Found {len(img_files)} images")

        images = []
        for idx, (num, img_file) in enumerate(img_files):
            img_path = img_folder / img_file
            try:
                img = plt.imread(img_path)
                img = tf.image.resize(img, self.img_size).numpy()
                if len(img.shape) == 2:
                    img = np.stack([img, img, img], axis=-1)
                elif img.shape[-1] == 4:
                    img = img[:, :, :3]
                images.append(img)

                if (idx + 1) % 500 == 0:
                    print(f"    Progress: {idx+1}/{len(img_files)}")
            except Exception as e:
                print(f"  Error loading {img_file}: {e}")
                continue

        images = np.array(images, dtype=np.float32)
        print(f"  Loaded {len(images)} images")
        return images

    def preprocess_data(self, images, annotations):
        """Preprocess images and extract labels"""
        min_len = min(len(images), len(annotations))
        images = images[:min_len]
        annotations = annotations[:min_len]

        # Normalize
        images = images.astype('float32') / 255.0

        # Extract labels
        expressions = annotations[:, 0].astype(int)
        valence = annotations[:, 1].astype(float)
        arousal = annotations[:, 2].astype(float)

        # Filter valid
        valid_idx = (valence != -2) & (arousal != -2)
        print(f"  Valid samples: {np.sum(valid_idx)}/{len(valid_idx)}")

        return (images[valid_idx], expressions[valid_idx],
                valence[valid_idx], arousal[valid_idx])


# ============================================================================
# MODEL ARCHITECTURE MODULE
# ============================================================================

class CNNModels:
    """Factory class for creating CNN models"""

    def __init__(self, input_shape=(224, 224, 3), num_classes=8):
        self.input_shape = input_shape
        self.num_classes = num_classes

    def build_resnet50(self, trainable_layers=30):
        """Build ResNet50-based model (optimized for 4GB GPU)"""
        base_model = ResNet50(
            include_top=False,
            weights='imagenet',
            input_shape=self.input_shape
        )

        # Freeze more layers for memory efficiency
        for layer in base_model.layers[:-trainable_layers]:
            layer.trainable = False

        x = base_model.output
        x = layers.GlobalAveragePooling2D()(x)
        x = layers.Dense(256, activation='relu')(x)
        x = layers.Dropout(0.4)(x)
        x = layers.Dense(128, activation='relu')(x)
        x = layers.Dropout(0.3)(x)

        # Output heads
        expr_out = layers.Dense(self.num_classes, activation='softmax',
                               name='expression', dtype='float32')(x)
        val_out = layers.Dense(1, activation='tanh',
                              name='valence', dtype='float32')(x)
        ar_out = layers.Dense(1, activation='tanh',
                             name='arousal', dtype='float32')(x)

        model = models.Model(
            inputs=base_model.input,
            outputs=[expr_out, val_out, ar_out],
            name='ResNet50_MultiTask'
        )

        return model

    def build_efficientnet(self, trainable_layers=30):
        """Build EfficientNetB0-based model"""
        base_model = EfficientNetB0(
            include_top=False,
            weights='imagenet',
            input_shape=self.input_shape
        )

        for layer in base_model.layers[:-trainable_layers]:
            layer.trainable = False

        x = base_model.output
        x = layers.GlobalAveragePooling2D()(x)
        x = layers.Dense(256, activation='relu')(x)
        x = layers.Dropout(0.4)(x)
        x = layers.Dense(128, activation='relu')(x)
        x = layers.Dropout(0.3)(x)

        expr_out = layers.Dense(self.num_classes, activation='softmax',
                               name='expression', dtype='float32')(x)
        val_out = layers.Dense(1, activation='tanh',
                              name='valence', dtype='float32')(x)
        ar_out = layers.Dense(1, activation='tanh',
                             name='arousal', dtype='float32')(x)

        model = models.Model(
            inputs=base_model.input,
            outputs=[expr_out, val_out, ar_out],
            name='EfficientNetB0_MultiTask'
        )

        return model


# ============================================================================
# TRAINING MODULE
# ============================================================================

class ModelTrainer:
    """Handles model training"""

    def __init__(self, model, model_name='model'):
        self.model = model
        self.model_name = model_name
        self.history = None

    def compile_model(self, learning_rate=0.0001):
        """Compile model"""
        optimizer = optimizers.Adam(learning_rate=learning_rate)

        self.model.compile(
            optimizer=optimizer,
            loss={
                'expression': 'sparse_categorical_crossentropy',
                'valence': 'mse',
                'arousal': 'mse'
            },
            loss_weights={
                'expression': 1.0,
                'valence': 0.5,
                'arousal': 0.5
            },
            metrics={
                'expression': ['accuracy'],
                'valence': ['mae'],
                'arousal': ['mae']
            }
        )

    def get_callbacks(self):
        """Setup callbacks"""
        return [
            EarlyStopping(
                monitor='val_loss',
                patience=8,
                restore_best_weights=True,
                verbose=1
            ),
            ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=4,
                min_lr=1e-7,
                verbose=1
            ),
            ModelCheckpoint(
                f'{self.model_name}_best.h5',
                monitor='val_loss',
                save_best_only=True,
                verbose=0
            )
        ]

    def train(self, train_data, val_data, epochs=20):
        """Train the model"""
        print(f"\nTraining {self.model_name}...")

        self.history = self.model.fit(
            train_data,
            validation_data=val_data,
            epochs=epochs,
            callbacks=self.get_callbacks(),
            verbose=1
        )

        return self.history


# ============================================================================
# EVALUATION METRICS MODULE
# ============================================================================

class MetricsCalculator:
    """Calculate evaluation metrics"""

    @staticmethod
    def categorical_metrics(y_true, y_pred, y_pred_proba=None):
        """Categorical metrics"""
        metrics = {}
        metrics['accuracy'] = accuracy_score(y_true, y_pred)
        metrics['f1_score'] = f1_score(y_true, y_pred, average='weighted')
        metrics['cohens_kappa'] = cohen_kappa_score(y_true, y_pred)

        # Krippendorff's Alpha (simplified)
        try:
            from sklearn.metrics import cohen_kappa_score as kappa
            metrics['krippendorfs_alpha'] = kappa(y_true, y_pred)
        except:
            metrics['krippendorfs_alpha'] = metrics['cohens_kappa']

        if y_pred_proba is not None:
            try:
                metrics['auc_roc'] = roc_auc_score(
                    y_true, y_pred_proba,
                    multi_class='ovr', average='weighted'
                )

                pr_aucs = []
                for i in range(y_pred_proba.shape[1]):
                    y_bin = (y_true == i).astype(int)
                    p, r, _ = precision_recall_curve(y_bin, y_pred_proba[:, i])
                    pr_aucs.append(auc(r, p))
                metrics['auc_pr'] = np.mean(pr_aucs)
            except:
                metrics['auc_roc'] = None
                metrics['auc_pr'] = None

        return metrics

    @staticmethod
    def continuous_metrics(y_true, y_pred):
        """Continuous metrics"""
        metrics = {}
        metrics['rmse'] = np.sqrt(np.mean((y_true - y_pred) ** 2))
        metrics['correlation'], _ = pearsonr(y_true.flatten(), y_pred.flatten())

        # SAGR
        correct_signs = np.sign(y_true) == np.sign(y_pred)
        metrics['sagr'] = np.mean(correct_signs)

        # CCC
        mean_true = np.mean(y_true)
        mean_pred = np.mean(y_pred)
        var_true = np.var(y_true)
        var_pred = np.var(y_pred)
        cov = np.mean((y_true - mean_true) * (y_pred - mean_pred))

        ccc_num = 2 * cov
        ccc_den = var_true + var_pred + (mean_true - mean_pred) ** 2
        metrics['ccc'] = ccc_num / ccc_den if ccc_den != 0 else 0

        return metrics


# ============================================================================
# VISUALIZATION MODULE
# ============================================================================

class Visualizer:
    """Visualization functions"""

    @staticmethod
    def plot_training_history(history, model_name='Model'):
        """Plot training history"""
        fig, axes = plt.subplots(2, 2, figsize=(15, 12))

        axes[0, 0].plot(history.history['loss'], label='Train')
        axes[0, 0].plot(history.history['val_loss'], label='Val')
        axes[0, 0].set_title(f'{model_name} - Total Loss')
        axes[0, 0].set_xlabel('Epoch')
        axes[0, 0].set_ylabel('Loss')
        axes[0, 0].legend()
        axes[0, 0].grid(True)

        axes[0, 1].plot(history.history['expression_accuracy'], label='Train')
        axes[0, 1].plot(history.history['val_expression_accuracy'], label='Val')
        axes[0, 1].set_title(f'{model_name} - Expression Accuracy')
        axes[0, 1].set_xlabel('Epoch')
        axes[0, 1].set_ylabel('Accuracy')
        axes[0, 1].legend()
        axes[0, 1].grid(True)

        axes[1, 0].plot(history.history['valence_mae'], label='Train')
        axes[1, 0].plot(history.history['val_valence_mae'], label='Val')
        axes[1, 0].set_title(f'{model_name} - Valence MAE')
        axes[1, 0].set_xlabel('Epoch')
        axes[1, 0].set_ylabel('MAE')
        axes[1, 0].legend()
        axes[1, 0].grid(True)

        axes[1, 1].plot(history.history['arousal_mae'], label='Train')
        axes[1, 1].plot(history.history['val_arousal_mae'], label='Val')
        axes[1, 1].set_title(f'{model_name} - Arousal MAE')
        axes[1, 1].set_xlabel('Epoch')
        axes[1, 1].set_ylabel('MAE')
        axes[1, 1].legend()
        axes[1, 1].grid(True)

        plt.tight_layout()
        plt.savefig(f'{model_name}_training.png', dpi=150, bbox_inches='tight')
        plt.show()

    @staticmethod
    def plot_confusion_matrix(y_true, y_pred, labels, model_name='Model'):
        """Plot confusion matrix"""
        cm = confusion_matrix(y_true, y_pred)

        plt.figure(figsize=(10, 8))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=labels.values(),
                   yticklabels=labels.values())
        plt.title(f'{model_name} - Confusion Matrix')
        plt.ylabel('True')
        plt.xlabel('Predicted')
        plt.savefig(f'{model_name}_confusion.png', dpi=150, bbox_inches='tight')
        plt.show()

    @staticmethod
    def plot_samples(images, y_true, y_pred, labels, n=8):
        """Plot sample predictions"""
        fig, axes = plt.subplots(2, 4, figsize=(16, 8))
        axes = axes.ravel()

        indices = np.random.choice(len(images), n, replace=False)

        for idx, ax_idx in enumerate(indices):
            ax = axes[idx]
            ax.imshow(images[ax_idx])

            true_label = labels[y_true[ax_idx]]
            pred_label = labels[y_pred[ax_idx]]
            color = 'green' if y_true[ax_idx] == y_pred[ax_idx] else 'red'

            ax.set_title(f"True: {true_label}\nPred: {pred_label}",
                        color=color, fontsize=10)
            ax.axis('off')

        plt.tight_layout()
        plt.savefig('sample_predictions.png', dpi=150, bbox_inches='tight')
        plt.show()


# ============================================================================
# MAIN EXECUTION
# ============================================================================

def main():
    """Main pipeline"""

    print("\n" + "="*80)
    print("FACIAL EXPRESSION RECOGNITION WITH CNN")
    print("="*80)

    # Dataset path
    data_path = data_path = Path('/content/Dataset/Dataset/Dataset')
# Load ALL data first (no split parameter)
    # Load ALL data first (no split parameter)
    print("\n[STEP 1] Loading Data...")
    loader = DataLoader(data_path)

    print("\nLoading all images...")
    all_images = loader.load_images('train')  # Load all

    print("\nLoading all annotations...")
    all_annotations = loader.load_annotations('train')  # Load all

    # Preprocess
    all_images, all_expr, all_val, all_ar = loader.preprocess_data(
        all_images, all_annotations
    )

    print(f"\nTotal valid samples: {len(all_images)}")

    # Now split into train/val/test (80/10/10)
    n_total = len(all_images)
    n_train = int(0.8 * n_total)
    n_val = int(0.1 * n_total)

    train_images = all_images[:n_train]
    train_expr = all_expr[:n_train]
    train_val = all_val[:n_train]
    train_ar = all_ar[:n_train]

    val_images = all_images[n_train:n_train+n_val]
    val_expr = all_expr[n_train:n_train+n_val]
    val_val = all_val[n_train:n_train+n_val]
    val_ar = all_ar[n_train:n_train+n_val]

    test_images = all_images[n_train+n_val:]
    test_expr = all_expr[n_train+n_val:]
    test_val = all_val[n_train+n_val:]
    test_ar = all_ar[n_train+n_val:]

    print(f"\n✓ Data split:")
    print(f"  Training: {len(train_images)}")
    print(f"  Validation: {len(val_images)}")
    print(f"  Test: {len(test_images)}")

    if len(train_images) == 0:
        print("\n⚠️ ERROR: No training data!")
        return

    # Create datasets (optimized batch size for 4GB GPU)
    print("\n[STEP 2] Creating Datasets...")
    batch_size = 16  # Smaller for 4GB GPU

    train_dataset = tf.data.Dataset.from_tensor_slices((
        train_images,
        {
            'expression': train_expr,
            'valence': train_val.reshape(-1, 1),
            'arousal': train_ar.reshape(-1, 1)
        }
    )).shuffle(1000).batch(batch_size).prefetch(tf.data.AUTOTUNE)

    val_dataset = tf.data.Dataset.from_tensor_slices((
        val_images,
        {
            'expression': val_expr,
            'valence': val_val.reshape(-1, 1),
            'arousal': val_ar.reshape(-1, 1)
        }
    )).batch(batch_size).prefetch(tf.data.AUTOTUNE)

    test_dataset = tf.data.Dataset.from_tensor_slices((
        test_images,
        {
            'expression': test_expr,
            'valence': test_val.reshape(-1, 1),
            'arousal': test_ar.reshape(-1, 1)
        }
    )).batch(batch_size).prefetch(tf.data.AUTOTUNE)

    print(f"✓ Datasets created (batch size: {batch_size})")

    # Train models
    results = {}
    model_builder = CNNModels(num_classes=8)

    # ResNet50
    print("\n[STEP 3] Training ResNet50...")
    resnet_model = model_builder.build_resnet50(trainable_layers=30)
    resnet_trainer = ModelTrainer(resnet_model, 'ResNet50')
    resnet_trainer.compile_model(learning_rate=0.0001)

    print(f"Parameters: {resnet_model.count_params():,}")
    resnet_history = resnet_trainer.train(train_dataset, val_dataset, epochs=20)

    # EfficientNet
    print("\n[STEP 4] Training EfficientNetB0...")
    effnet_model = model_builder.build_efficientnet(trainable_layers=30)
    effnet_trainer = ModelTrainer(effnet_model, 'EfficientNetB0')
    effnet_trainer.compile_model(learning_rate=0.0001)

    print(f"Parameters: {effnet_model.count_params():,}")
    effnet_history = effnet_trainer.train(train_dataset, val_dataset, epochs=20)

    # Evaluate
    print("\n[STEP 5] Evaluating Models...")
    calculator = MetricsCalculator()
    visualizer = Visualizer()

    # ResNet50 evaluation
    print("\nEvaluating ResNet50...")
    resnet_pred = resnet_model.predict(test_images, batch_size=batch_size)
    resnet_expr_pred = np.argmax(resnet_pred[0], axis=1)
    resnet_val_pred = resnet_pred[1].flatten()
    resnet_ar_pred = resnet_pred[2].flatten()

    results['ResNet50'] = {
        'categorical': calculator.categorical_metrics(
            test_expr, resnet_expr_pred, resnet_pred[0]
        ),
        'valence': calculator.continuous_metrics(test_val, resnet_val_pred),
        'arousal': calculator.continuous_metrics(test_ar, resnet_ar_pred)
    }

    # EfficientNet evaluation
    print("Evaluating EfficientNetB0...")
    effnet_pred = effnet_model.predict(test_images, batch_size=batch_size)
    effnet_expr_pred = np.argmax(effnet_pred[0], axis=1)
    effnet_val_pred = effnet_pred[1].flatten()
    effnet_ar_pred = effnet_pred[2].flatten()

    results['EfficientNetB0'] = {
        'categorical': calculator.categorical_metrics(
            test_expr, effnet_expr_pred, effnet_pred[0]
        ),
        'valence': calculator.continuous_metrics(test_val, effnet_val_pred),
        'arousal': calculator.continuous_metrics(test_ar, effnet_ar_pred)
    }

    # Visualizations
    print("\n[STEP 6] Creating Visualizations...")
    visualizer.plot_training_history(resnet_history, 'ResNet50')
    visualizer.plot_training_history(effnet_history, 'EfficientNetB0')
    visualizer.plot_confusion_matrix(
        test_expr, resnet_expr_pred, loader.emotion_labels, 'ResNet50'
    )
    visualizer.plot_confusion_matrix(
        test_expr, effnet_expr_pred, loader.emotion_labels, 'EfficientNetB0'
    )
    visualizer.plot_samples(
        test_images, test_expr, resnet_expr_pred, loader.emotion_labels
    )

    # Print results
    print("\n" + "="*80)
    print("RESULTS SUMMARY")
    print("="*80)

    for model_name, metrics in results.items():
        print(f"\n{model_name}:")
        print(f"\nExpression Metrics:")
        for k, v in metrics['categorical'].items():
            if v is not None:
                print(f"  {k}: {v:.4f}")

        print(f"\nValence Metrics:")
        for k, v in metrics['valence'].items():
            print(f"  {k}: {v:.4f}")

        print(f"\nArousal Metrics:")
        for k, v in metrics['arousal'].items():
            print(f"  {k}: {v:.4f}")

    # Comparison
    print("\n" + "="*80)
    print("MODEL COMPARISON")
    print("="*80)

    comparison = pd.DataFrame({
        'ResNet50': [
            results['ResNet50']['categorical']['accuracy'],
            results['ResNet50']['categorical']['f1_score'],
            results['ResNet50']['valence']['rmse'],
            results['ResNet50']['arousal']['rmse']
        ],
        'EfficientNetB0': [
            results['EfficientNetB0']['categorical']['accuracy'],
            results['EfficientNetB0']['categorical']['f1_score'],
            results['EfficientNetB0']['valence']['rmse'],
            results['EfficientNetB0']['arousal']['rmse']
        ]
    }, index=['Expr Accuracy', 'F1-Score', 'Valence RMSE', 'Arousal RMSE'])

    print(comparison)

    print("\n✓ Training complete!")
    print("\nNext steps:")
    print("1. Upload this notebook to GitHub")
    print("2. Create report PDF with results")
    print("3. Submit on LMS")

if __name__ == "__main__":
    main()
```
